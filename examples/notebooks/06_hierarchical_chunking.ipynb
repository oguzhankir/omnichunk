# Hierarchical chunking

Multi-level `ChunkTree`: leaves for embeddings, roots for LLM context.

Install the package in this notebook kernel before running: `%pip install omnichunk[code]`.
To test a checkout, install its absolute path with `%pip install -e /path/to/omnichunk[code]`.


```python
import omnichunk
from omnichunk import Chunker

print("omnichunk", omnichunk.__version__, "from", omnichunk.__file__)

src = "def f():\n    return 1\n" * 5
c = Chunker(size_unit="chars", min_chunk_size=1)
tree = c.hierarchical_chunk("m.py", src, levels=[48, 200, 800])
print("levels", tree.level_count, "nodes", len(tree.nodes))
print("leaves", len(tree.leaves()), "roots", len(tree.roots()))
```

```python
def ascii_tree(tree):
    def visit(chunk, depth=0):
        print(f"{'  ' * depth}bytes {chunk.byte_range.start}-{chunk.byte_range.end}")
        for child in tree.children(chunk):
            visit(child, depth + 1)

    for root in tree.roots():
        visit(root)


ascii_tree(tree)
```
